Part 15 — Text Analysis Using NumPy

```python
# Create a sentence
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"

# Convert the sentence to lowercase and split it into words
tokens2 = sentence2.lower().split()

# Display the tokens
print(tokens2)
```

From text to tokens

```python
# Convert tokens into a NumPy array
tokens2_array = np.array(tokens2)

# Find the unique words
vocabulary2 = np.unique(tokens2_array)

# Display the vocabulary
print("Vocabulary:", vocabulary2)

# Display the number of unique words
print("Vocabulary size:", len(vocabulary2))
```

Building a Vocabulary

```python
# Create a function to generate a word-count vector
def bag_of_words_vector(tokens, vocabulary):

    # Count how many times each vocabulary word appears
    vector = np.array([np.sum(np.array(tokens) == word) for word in vocabulary])

    # Return the word-count vector
    return vector
```

```python
# Create the Bag-of-Words vector
vec2 = bag_of_words_vector(tokens2, vocabulary2)

# Display the vocabulary
print("Vocabulary:", vocabulary2)

# Display the word-count vector
print("Word-count vector:", vec2)
```

Word Frequency Vector

```python
# Create the first sentence
sentence_a = "I like machine learning"

# Create the second similar sentence
sentence_b = "I like machine learning algorithms"

# Tokenize both sentences
tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()

# Create a shared vocabulary
combined_vocab = np.unique(np.array(tokens_a + tokens_b))

# Convert both sentences into vectors
vec_a = bag_of_words_vector(tokens_a, combined_vocab)
vec_b = bag_of_words_vector(tokens_b, combined_vocab)

# Calculate cosine similarity
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))

# Display the similarity
print("Cosine similarity:", cosine_sim)
```

Comparing Two Sentences

```python
# Create the first sentence
sentence_a = "I like machine learning"

# Create the second different sentence
sentence_b = "The weather is very hot today"

# Tokenize both sentences
tokens_a = sentence_a.lower().split()
tokens_b = sentence_b.lower().split()

# Create a shared vocabulary
combined_vocab = np.unique(np.array(tokens_a + tokens_b))

# Convert both sentences into vectors
vec_a = bag_of_words_vector(tokens_a, combined_vocab)
vec_b = bag_of_words_vector(tokens_b, combined_vocab)

# Calculate cosine similarity
cosine_sim = np.dot(vec_a, vec_b) / (np.linalg.norm(vec_a) * np.linalg.norm(vec_b))

# Display the similarity
print("Cosine similarity:", cosine_sim)
```

Most Frequent Words

```python
# Create a paragraph
paragraph = "Python is useful. Python is easy to learn. Python is popular in data science. Data science uses Python."

# Convert the paragraph to lowercase and split it into words
tokens = paragraph.lower().replace(".", "").split()

# Convert tokens into a NumPy array
tokens_array = np.array(tokens)

# Create the vocabulary
vocabulary = np.unique(tokens_array)

# Create the word-count vector
vector = bag_of_words_vector(tokens, vocabulary)

# Sort the indices according to word frequency
indices = np.argsort(vector)[::-1]

# Select the top 3 most frequent words
top_3 = vocabulary[indices[:3]]

# Display the top 3 words
print("Top 3 most frequent words:", top_3)
```

Mini Project:Simple Text Similarity Checker

```python
# Create three sentences
sentence1 = "I like machine learning"
sentence2 = "I enjoy machine learning"
sentence3 = "The weather is very hot today"

# Tokenize all three sentences
tokens1 = sentence1.lower().split()
tokens2 = sentence2.lower().split()
tokens3 = sentence3.lower().split()

# Create one shared vocabulary
combined_vocab = np.unique(
    np.array(tokens1 + tokens2 + tokens3)
)

# Convert each sentence into a Bag-of-Words vector
vec1 = bag_of_words_vector(tokens1, combined_vocab)
vec2 = bag_of_words_vector(tokens2, combined_vocab)
vec3 = bag_of_words_vector(tokens3, combined_vocab)

# Calculate similarity between sentence 1 and sentence 2
similarity_12 = np.dot(vec1, vec2) / (
    np.linalg.norm(vec1) * np.linalg.norm(vec2)
)

# Calculate similarity between sentence 1 and sentence 3
similarity_13 = np.dot(vec1, vec3) / (
    np.linalg.norm(vec1) * np.linalg.norm(vec3)
)

# Calculate similarity between sentence 2 and sentence 3
similarity_23 = np.dot(vec2, vec3) / (
    np.linalg.norm(vec2) * np.linalg.norm(vec3)
)

# Display the similarities
print("Similarity 1-2:", similarity_12)
print("Similarity 1-3:", similarity_13)
print("Similarity 2-3:", similarity_23)

# Find the highest similarity
similarities = np.array([
    similarity_12,
    similarity_13,
    similarity_23
])

# Find the index of the highest similarity
best_pair = np.argmax(similarities)

# Display the most similar pair
if best_pair == 0:
    print("Most similar: Sentence 1 and Sentence 2")
elif best_pair == 1:
    print("Most similar: Sentence 1 and Sentence 3")
else:
    print("Most similar: Sentence 2 and Sentence 3")
```
