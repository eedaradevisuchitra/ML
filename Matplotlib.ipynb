```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("/content/Student_Performance_Dataset.csv")

data.head()
```

```python
data.shape
```

```python
data.columns
```

```python
data.describe()            #AP24110011392
```

```python
x_val = [1, 2, 3, 4, 5]
y_val = [50, 60, 70, 65, 80]

plt.plot(x_val, y_val)
plt.xlabel("Student")
plt.ylabel("Final Percentage")
plt.title("Student Performance")
plt.show()
```

```python
performance_count = data['Performance_Level'].value_counts()

x_bar = performance_count.index
y_bar = performance_count

plt.figure(figsize=(8,6))                 #AP24110011392

plt.title("Students by Performance Level")
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.bar(x_bar, y_bar, width=0.5, color='orange')

plt.title("Students by Performance Level", fontsize=15)
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")

plt.show()
```

```python
sns.countplot(x='Performance_Level', data=data,
              order=performance_count.index)

plt.title("Performance Level Count")
plt.xlabel("Performance Level")
plt.ylabel("Number of Students")

plt.show()
```

```python
gender_data = data['Gender'].value_counts()

plt.pie(gender_data,
        labels=gender_data.index,
        autopct='%.2f%%')

plt.title("Gender Distribution")
plt.show()
```

```python
pass_data = data['Pass_Fail'].value_counts()

plt.pie(pass_data,
        labels=pass_data.index,
        autopct='%.2f%%',                     #AP24110011392

        explode=(0, 0.1))

plt.title("Pass and Fail Distribution")
plt.show()
```

```python
plt.hist(data['Age'], bins=10)

plt.xlabel("Age")                     #AP24110011392

plt.ylabel("Number of Students")
plt.title("Age Distribution")

plt.show()
```

```python
plt.hist(data['Final_Percentage'], bins=10)

plt.xlabel("Final Percentage")
plt.ylabel("Number of Students")
plt.title("Final Percentage Distribution")

plt.show()
```

```python
sns.histplot(data['Final_Percentage'], bins=10)

plt.title("Final Percentage Distribution")
plt.xlabel("Final Percentage")
plt.ylabel("Count")

plt.show()
```

```python
sns.kdeplot(data['Final_Percentage'])

plt.title("Final Percentage Distribution")
plt.xlabel("Final Percentage")

plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.boxplot(y=data['Final_Percentage'])

plt.ylabel("Final Percentage")
plt.title("Final Percentage Box Plot")          #AP24110011392

plt.show()
```

```python
plt.scatter(data['Study_Hours_Per_Day'],
            data['Final_Percentage'])

plt.xlabel("Study Hours Per Day")
plt.ylabel("Final Percentage")
plt.title("Study Hours vs Final Percentage")

plt.show()
```

```python
plt.scatter(data['Attendance_Percentage'],
            data['Final_Percentage'])

plt.xlabel("Attendance Percentage")
plt.ylabel("Final Percentage")
plt.title("Attendance vs Final Percentage")

plt.show()
```

```python
x = ['Math', 'Science', 'English']

math = data['Math_Score'].mean()
science = data['Science_Score'].mean()
english = data['English_Score'].mean()

y = [math, science, english]

plt.plot(x, y, marker='o')

plt.xlabel("Subjects")
plt.ylabel("Average Score")
plt.title("Average Subject Performance")

plt.show()
```

```python
sns.countplot(x='Gender',
              hue='Performance_Level',
              data=data)

plt.title("Performance Level by Gender")
plt.show()
```

```python
numeric_data = data.select_dtypes(include=np.number)

plt.figure(figsize=(10,8))

sns.heatmap(numeric_data.corr(), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
